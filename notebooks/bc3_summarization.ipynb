# BC3 Summarization — Condition A vs Condition B

Displays the pre-generated summaries side by side for each of the 6 test threads.

- **Condition A** — raw email text only
- **Condition B** — email text + BERT+LoRA predicted speech act labels

Generated by `src/bc3_summarize.py` using Gemini 3.6 Flash.

```python
import os, sys
from pathlib import Path

# Works whether the notebook is opened from the repo root or from notebooks/
here = Path(os.path.abspath(''))
REPO_ROOT = here.parent if here.name == 'notebooks' else here
sys.path.insert(0, str(REPO_ROOT / 'src'))

import pandas as pd
from IPython.display import display, Markdown

df_a = pd.read_csv(REPO_ROOT / 'results' / 'bc3_summaries_condition_a.csv')
df_b = pd.read_csv(REPO_ROOT / 'results' / 'bc3_summaries_condition_b.csv')

print(f'Loaded {len(df_a)} threads for Condition A')
print(f'Loaded {len(df_b)} threads for Condition B')
```

```python
merged = df_a.merge(df_b[['listno', 'summary_b']], on='listno')

for i, row in merged.iterrows():
    display(Markdown(
        f"---\n"
        f"## [{i+1}/6] `{row['listno']}` — {row['thread_name']} ({int(row['n_emails'])} emails)\n\n"
        f"### Condition A — raw text\n\n{row['summary_a']}\n\n"
        f"### Condition B — with speech act labels\n\n{row['summary_b']}"
    ))
```
